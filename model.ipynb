```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load your dataset
data = pd.read_csv('/content/DataSet.csv')

# Create a countplot (good for categorical data)
plt.figure(figsize=(8, 6))
sns.countplot(x='Gender', hue='stress_level', data=data)
plt.title('Stress Levels by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Convert categorical columns to numerical values
data_encoded = data.copy()
for col in data_encoded.select_dtypes(include=['object', 'category']):
    data_encoded[col] = data_encoded[col].astype('category').cat.codes

# Calculate the correlation matrix
correlation_matrix = data_encoded.corr()

# Display the correlation matrix
print(correlation_matrix)


# Visualize the correlation matrix using a heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Attributes')
plt.show()
```

```python
def features_to_points(col_data):
    for i,text in enumerate(col_data):
        text = str(text)
        splitted_text = text.split(';')
        if "None" in splitted_text:
            col_data[i] = 0
        else:
            col_data[i] = len(splitted_text)
features_to_points(data['friends_issues'])
features_to_points(data['family_issues'])
features_to_points(data['financial_issues'])
features_to_points(data['health_issues'])
```

```python
data.head()
```

```python
def one_or_zero(col_data):
    for i,text in enumerate(col_data):
        if (text == 'Male') or (text=='Yes'):
            col_data[i] = 1
        else:
            col_data[i] = 0
one_or_zero(data['Gender'])
one_or_zero(data['overload'])
one_or_zero(data['unpleasant'])
one_or_zero(data['academic'])
one_or_zero(data['career'])
one_or_zero(data['criticism'])
one_or_zero(data['Conflicts'])
```

```python
data.head()
```

```python
data['stress_level']=data['stress_level'].replace({'Acute Stress':0,
'Episodic acute stress':1,
'Chronic stress': 2,'No Stress': 3})
```

```python
data['stress_level'].unique()
```

```python
X = data.drop('stress_level', axis=1)
y = data['stress_level']
```

```python
X
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn import metrics

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
svm_model = SVC()
svm_model.fit(X_train, y_train)
y_pred = svm_model.predict(X_test)
print("SVM Accuracy:", metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier()
dt_model.fit(X_train, y_train)
y_pred = dt_model.predict(X_test)
print("Decision Tree Accuracy:", metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn.linear_model import LogisticRegression

log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train, y_train)
y_pred = log_model.predict(X_test)
print("Logistic Regression Accuracy:", metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier()
knn_model.fit(X_train, y_train)
y_pred = knn_model.predict(X_test)
print("KNN Accuracy:", metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn.naive_bayes import GaussianNB

nb_model = GaussianNB()
nb_model.fit(X_train, y_train)
y_pred = nb_model.predict(X_test)
print("Naive Bayes Accuracy:", metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn.ensemble import GradientBoostingClassifier

gb_model = GradientBoostingClassifier()
gb_model.fit(X_train, y_train)
y_pred = gb_model.predict(X_test)
print("Gradient Boosting Accuracy:", metrics.accuracy_score(y_test, y_pred))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import joblib

# Split your data
X = data.drop("stress_level", axis=1)
y = data["stress_level"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Train model
clf = RandomForestClassifier()
clf.fit(X_train, y_train)

# Predict and evaluate
y_pred = clf.predict(X_test)
print("Random Accuracy:", accuracy_score(y_test, y_pred))

# Save the trained model
joblib.dump(clf, 'random_forest_model.pkl')
```

```python
import joblib
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Load the best saved model
best_model = joblib.load("random_forest_model.pkl")

# Predict using test data
y_pred = best_model.predict(X_test)

# Output results
print("Accuracy on Test Data:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```
